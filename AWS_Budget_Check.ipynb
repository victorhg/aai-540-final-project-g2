```python
import boto3
import sagemaker
from datetime import datetime, timezone

sess = sagemaker.Session()
region = boto3.Session().region_name
sm = boto3.client("sagemaker", region_name=region)

print("Region:", region)
print("Time:", datetime.now(timezone.utc).isoformat())
print("="*80)

# 1) Endpoints that are actively billable
eps = sm.list_endpoints(MaxResults=100)["Endpoints"]
live = [e for e in eps if e["EndpointStatus"] in ("InService", "Updating", "Creating")]
print(f"\nEndpoints (billable if InService): {len(live)}")
for e in live:
    print(" -", e["EndpointName"], e["EndpointStatus"])

# 2) Monitoring schedules that can keep spawning processing jobs
scheds = sm.list_monitoring_schedules(MaxResults=100).get("MonitoringScheduleSummaries", [])
active_scheds = [s for s in scheds if s["MonitoringScheduleStatus"] in ("Scheduled", "Pending", "InProgress")]
print(f"\nMonitoring schedules (potentially active): {len(active_scheds)}")
for s in active_scheds:
    print(" -", s["MonitoringScheduleName"], s["MonitoringScheduleStatus"])

# 3) Processing jobs currently running (Model Monitor runs show up here)
pj = sm.list_processing_jobs(MaxResults=100).get("ProcessingJobSummaries", [])
running_pj = [p for p in pj if p["ProcessingJobStatus"] in ("InProgress", "Stopping")]
print(f"\nProcessing jobs running: {len(running_pj)}")
for p in running_pj[:30]:
    print(" -", p["ProcessingJobName"], p["ProcessingJobStatus"])

# 4) Transform jobs running (batch transform)
tj = sm.list_transform_jobs(MaxResults=100).get("TransformJobSummaries", [])
running_tj = [t for t in tj if t["TransformJobStatus"] in ("InProgress", "Stopping")]
print(f"\nTransform jobs running: {len(running_tj)}")
for t in running_tj[:30]:
    print(" -", t["TransformJobName"], t["TransformJobStatus"])

# 5) Training jobs running (rare in your script since you wait for completion)
tr = sm.list_training_jobs(MaxResults=100).get("TrainingJobSummaries", [])
running_tr = [t for t in tr if t["TrainingJobStatus"] in ("InProgress", "Stopping")]
print(f"\nTraining jobs running: {len(running_tr)}")
for t in running_tr[:30]:
    print(" -", t["TrainingJobName"], t["TrainingJobStatus"])
```
